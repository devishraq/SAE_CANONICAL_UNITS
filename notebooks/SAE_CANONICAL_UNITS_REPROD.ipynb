## Setup

```python
import os, shutil
from huggingface_hub import login
from google.colab import userdata

%cd /content

REPO_PATH = "/content/SAE_CANONICAL_UNITS"

if os.path.exists(REPO_PATH):
    print(f"Removing old {REPO_PATH}...")
    shutil.rmtree(REPO_PATH)

!git clone https://github.com/devishraq/SAE_CANONICAL_UNITS.git {REPO_PATH}
%cd {REPO_PATH}

!pip install -q -e .
!pip install -q -U nnterp sae-lens datasets

# --- Authentication Setup ---
try:
    hf_token = userdata.get('HF_TOKEN')
    os.environ["HF_TOKEN"] = hf_token
    login(token=hf_token)
    print("✅ HuggingFace login successful! (Gemma will work)")
except Exception:
    print("⚠️ HF_TOKEN not found in Colab secrets. Add it if you want to run Gemma.")

try:
    ndif_key = userdata.get('NDIF_API_KEY')
    os.environ["NDIF_API_KEY"] = ndif_key
    print("✅ NDIF API Key loaded! (Llama-3.1-8B remote execution will work)")
except Exception:
    print("⚠️ NDIF_API_KEY not found in Colab secrets. Add it if you want to run Llama-3.1-8B.")

print("\nReady")
```
